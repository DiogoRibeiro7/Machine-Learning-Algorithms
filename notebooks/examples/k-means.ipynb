# k-means clustering

A maintained example with reproducible k-means++ initialization.

```python
import numpy as np

from ml_algorithms import KMeans
```

```python
rng = np.random.default_rng(42)
a = rng.normal(loc=(-2.0, -1.0), scale=0.5, size=(80, 2))
b = rng.normal(loc=(2.0, 1.5), scale=0.6, size=(80, 2))
c = rng.normal(loc=(0.0, 3.5), scale=0.4, size=(80, 2))
X = np.vstack((a, b, c))
```

```python
model = KMeans(n_clusters=3, random_state=42).fit(X)

print(model.cluster_centers_)
print(model.inertia_)
print(model.converged_, model.n_iter_)
```
