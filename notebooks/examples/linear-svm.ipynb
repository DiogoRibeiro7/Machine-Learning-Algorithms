# Linear support-vector machine

A binary linear SVM using the maintained squared-hinge implementation.

```python
import numpy as np

from ml_algorithms import LinearSVM
```

```python
rng = np.random.default_rng(42)
negative = rng.normal(loc=(-1.5, -1.0), scale=0.6, size=(80, 2))
positive = rng.normal(loc=(1.5, 1.0), scale=0.6, size=(80, 2))
X = np.vstack((negative, positive))
y = np.array([0] * 80 + [1] * 80)
```

```python
model = LinearSVM(C=2.0).fit(X, y)

print(model.coef_)
print(model.intercept_)
print(model.converged_, model.n_iter_)
print(model.loss_)
```
