# Principal component analysis

A maintained PCA example using centered SVD.

```python
import numpy as np

from ml_algorithms import PCA
```

```python
rng = np.random.default_rng(42)
latent = rng.normal(size=(200, 2))
mixing = np.array([[2.0, 0.2], [0.8, 1.5], [-1.0, 0.4]])
X = latent @ mixing.T + rng.normal(scale=0.1, size=(200, 3))
```

```python
pca = PCA(n_components=2).fit(X)
Z = pca.transform(X)

print(pca.components_)
print(pca.explained_variance_ratio_)
print(Z.shape)
```
