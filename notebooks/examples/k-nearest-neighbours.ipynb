# k-nearest neighbours

A compact example using the maintained brute-force classifier.

```python
import numpy as np

from ml_algorithms import KNeighborsClassifier
```

```python
rng = np.random.default_rng(42)
class_a = rng.normal(loc=(-1.5, -1.0), scale=0.6, size=(50, 2))
class_b = rng.normal(loc=(1.5, 1.0), scale=0.6, size=(50, 2))
X = np.vstack((class_a, class_b))
y = np.array([0] * 50 + [1] * 50)
```

```python
model = KNeighborsClassifier(n_neighbors=5).fit(X, y)
queries = np.array([[-1.0, -0.5], [1.0, 0.5], [0.0, 0.0]])

print(model.predict(queries))
print(model.predict_proba(queries))
```
