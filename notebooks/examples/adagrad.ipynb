# AdaGrad on an anisotropic quadratic\n\nCompare adaptive coordinate scaling with ordinary fixed-rate gradient descent.

```python
import numpy as np\n\nfrom ml_algorithms.optimization import adagrad, gradient_descent\n
```

```python
matrix = np.array([[100.0, 0.0], [0.0, 1.0]])\n\ndef objective(theta: np.ndarray) -> float:\n    return float(0.5 * theta @ matrix @ theta)\n\ndef gradient(theta: np.ndarray) -> np.ndarray:\n    return matrix @ theta\n
```

```python
initial = np.array([10.0, 10.0])\n\nfixed = gradient_descent(\n    objective, gradient, initial, learning_rate=0.01, max_iter=200\n)\nadaptive = adagrad(\n    objective, gradient, initial, learning_rate=2.0, max_iter=200\n)\n\nprint(fixed.objective, fixed.parameters)\nprint(adaptive.objective, adaptive.parameters)\n
```
