# Gaussian Naive Bayes

A continuous-feature Naive Bayes example using the maintained package.

```python
import numpy as np

from ml_algorithms import GaussianNB
```

```python
rng = np.random.default_rng(42)
class_a = rng.normal(loc=(-1.0, 0.0), scale=(0.8, 0.5), size=(100, 2))
class_b = rng.normal(loc=(1.2, 1.0), scale=(0.6, 0.7), size=(100, 2))
X = np.vstack((class_a, class_b))
y = np.array([0] * 100 + [1] * 100)
```

```python
model = GaussianNB().fit(X, y)
queries = np.array([[-0.8, 0.1], [1.0, 0.9], [0.0, 0.5]])

print(model.predict(queries))
print(model.predict_proba(queries))
```
