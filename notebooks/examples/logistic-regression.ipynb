# Logistic regression

Binary logistic regression using the maintained package implementation.

```python
import numpy as np

from ml_algorithms import LogisticRegression
```

```python
rng = np.random.default_rng(42)
X = rng.normal(size=(200, 2))
logits = -0.4 + X @ np.array([1.8, -1.2])
probabilities = 1.0 / (1.0 + np.exp(-logits))
y = rng.binomial(1, probabilities)
```

```python
model = LogisticRegression(l2=0.1).fit(X, y)

print(model.intercept_)
print(model.coef_)
print(model.converged_, model.n_iter_)
```
