# Gradient descent

Minimize a two-dimensional convex quadratic with the reusable numerical utility.

```python
import numpy as np

from ml_algorithms.optimization import gradient_descent
```

```python
matrix = np.array([[4.0, 0.0], [0.0, 2.0]])
optimum = np.array([1.5, -2.0])

def objective(theta: np.ndarray) -> float:
    error = theta - optimum
    return float(0.5 * error @ matrix @ error)

def gradient(theta: np.ndarray) -> np.ndarray:
    return matrix @ (theta - optimum)
```

```python
result = gradient_descent(
    objective,
    gradient,
    [8.0, 5.0],
    learning_rate=0.2,
)

print(result.parameters)
print(result.objective)
print(result.converged, result.n_iter)
```
