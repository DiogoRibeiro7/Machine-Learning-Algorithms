# Linear regression from first principles

This example uses the maintained `LinearRegression` estimator from the package.

```python
import numpy as np

from ml_algorithms import LinearRegression
```

```python
rng = np.random.default_rng(42)
X = rng.normal(size=(100, 2))
true_coef = np.array([2.0, -1.5])
y = 0.75 + X @ true_coef + rng.normal(scale=0.1, size=100)
```

```python
model = LinearRegression().fit(X, y)

print(f"intercept: {model.intercept_:.3f}")
print(f"coefficients: {model.coef_}")
print(f"residual sum of squares: {model.residual_sum_squares_:.3f}")
```
